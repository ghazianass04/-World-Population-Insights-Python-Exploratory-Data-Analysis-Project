```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\saad\Desktop\PyProjects\world_population_data.csv")
```

```python
sns.histplot(df["2023 population"],bins=10,kde=True)
plt.title("Distribution of 2023 Population")
plt.xlabel("Population")
plt.ylabel("Frequency")
plt.show()
```

```python
df.groupby("continent")["2023 population"].mean().sort_values(ascending=False)
```

```python
top_10_countries=df.sort_values(by="2023 population", ascending=False).head(10)
top_10_countries
```

```python
sns.barplot(x="2023 population",y="country",data=top_10_countries)
plt.title("top 10 Countries by Population in 2023")
plt.xlabel("Population")
plt.ylabel("Country")
plt.show() 
```

```python
sns.scatterplot(x="2023 population",y="density (km²)",data=df)
plt.title("tPopulation and Density Scatter Plot")
plt.xlabel("2023 Population")
plt.ylabel("Density")
plt.show() 
```

```python
df.columns
```

```python
df.groupby("continent")["area (km²)"].sum().sort_values(ascending=False)
```

```python
sns.violinplot(x="continent",y="density (km²)",data=df,palette="viridis",hue="continent", legend=False)
plt.title("Population Density Distribution Across Continents")
plt.xlabel("Continent")
plt.ylabel("Population Density")
plt.show() 
```

```python
sns.boxplot(x="continent",y="density (km²)",data=df,palette="muted",hue="continent", showfliers=False)
plt.title("Population Density Distribution Across Continents (Without Outliers)")
plt.xlabel("Continent")
plt.ylabel("Population Density")
plt.show() 
```
